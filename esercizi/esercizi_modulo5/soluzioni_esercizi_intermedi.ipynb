## Esercizio 1.1: Creare e Attivare un Venv

**Obiettivo:** Creare un virtual environment per un nuovo progetto e installare una libreria.

**Nota:** Questo esercizio va eseguito da terminale, non nel notebook.


- Le librerie installate con `pip` vanno solo nel venv, non globalmente

## Esercizio 1.3: Installare Versioni Specifiche

**Obiettivo:** Installare versioni specifiche di librerie per evitare conflitti.

```bash
# Soluzione Esercizio 1.3 (da terminale)

# 1. Installa pandas versione esatta 1.5.0
pip install pandas==1.5.0

# 2. Installa numpy con versione minima 1.20.0
pip install "numpy>=1.20.0"

# 3. Verifica le versioni installate
pip show pandas
# Output:
# Name: pandas
# Version: 1.5.0
# ...

pip show numpy
# Output:
# Name: numpy
# Version: 1.26.3 (o superiore)
# ...

# 4. Crea requirements.txt con versioni esatte
pip freeze > requirements.txt

# Il file conterrà:
# numpy==1.26.3
# pandas==1.5.0
# python-dateutil==2.8.2
# pytz==2023.3.post1
# six==1.16.0
```

**Spiegazione:**
- `==` specifica una versione ESATTA (es. `pandas==1.5.0`)
- `>=` specifica una versione MINIMA (es. `numpy>=1.20.0`)
- `<=` specifica una versione MASSIMA
- Si possono combinare: `"pandas>=1.5.0,<2.0.0"` (versione 1.x)
- Le virgolette sono necessarie con `>`, `<` per evitare problemi con la shell

# 🎯 Sezione 2: File I/O

## Esercizio 2.1: Gestione Log File

**Obiettivo:** Creare un sistema di logging semplice che scrive su file.

````python
# Soluzione Esercizio 2.1

from datetime import datetime
from pathlib import Path

def scrivi_log(messaggio, livello="INFO"):
    """
    Scrive un messaggio di log nel file log.txt.
    Formato: [TIMESTAMP] LIVELLO: messaggio
    """
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    riga_log = f"[{timestamp}] {livello}: {messaggio}\n"
    
    # Modalità 'a' = append (aggiunge senza sovrascrivere)
    with open("log.txt", "a", encoding="utf-8") as f:
        f.write(riga_log)

# Test
scrivi_log("Applicazione avviata")
scrivi_log("Errore di connessione al database", livello="ERROR")
scrivi_log("Operazione completata con successo", livello="SUCCESS")
scrivi_log("Memoria al 85%", livello="WARNING")

# Leggi e mostra il log
print("\n=== Contenuto log.txt ===")
with open("log.txt", "r", encoding="utf-8") as f:
    print(f.read())

# Cleanup
Path("log.txt").unlink()
````

**Spiegazione:**
- `datetime.now().strftime()` formatta la data/ora corrente
- Modalità `'a'` (append) aggiunge al file senza cancellare il contenuto esistente
- `with open()` garantisce la chiusura automatica del file
- `encoding="utf-8"` gestisce correttamente caratteri accentati

## Esercizio 2.2: CSV di Prodotti

**Obiettivo:** Creare e leggere un file CSV con dati di prodotti.

````python
# Soluzione Esercizio 2.2

import csv
from pathlib import Path

# Parte 1: Scrivi i dati
prodotti = [
    {"nome": "Laptop", "prezzo": 999.99, "quantita": 5},
    {"nome": "Mouse", "prezzo": 29.99, "quantita": 20},
    {"nome": "Tastiera", "prezzo": 79.99, "quantita": 15}
]

# Scrivi CSV usando DictWriter
with open("prodotti.csv", "w", newline="", encoding="utf-8") as f:
    fieldnames = ["nome", "prezzo", "quantita"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    
    writer.writeheader()  # Scrive l'intestazione
    writer.writerows(prodotti)  # Scrive tutte le righe

print("✅ File prodotti.csv creato\n")

# Parte 2: Leggi e calcola
valore_totale_inventario = 0

with open("prodotti.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    
    print("📦 Inventario:")
    for riga in reader:
        nome = riga["nome"]
        prezzo = float(riga["prezzo"])
        quantita = int(riga["quantita"])
        
        valore_prodotto = prezzo * quantita
        valore_totale_inventario += valore_prodotto
        
        print(f"  {nome}: {quantita} x €{prezzo:.2f} = €{valore_prodotto:.2f}")

print(f"\n💰 Valore totale inventario: €{valore_totale_inventario:.2f}")

# Cleanup
Path("prodotti.csv").unlink()
````

**Spiegazione:**
- `newline=""` previene righe vuote extra su Windows
- `DictWriter` è più leggibile di `writer` quando hai molte colonne
- `writeheader()` scrive automaticamente i nomi delle colonne
- `DictReader` restituisce ogni riga come dizionario (chiave = nome colonna)
- I valori letti da CSV sono sempre stringhe, vanno convertiti con `int()` o `float()`

## Esercizio 2.3: Configurazione JSON

**Obiettivo:** Gestire un file di configurazione in JSON.

````python
# Soluzione Esercizio 2.3

import json
from pathlib import Path

# Crea un dizionario di configurazione
config = {
    "app_name": "MyApp",
    "version": "1.0.0",
    "database": {
        "host": "localhost",
        "port": 5432,
        "name": "mydb"
    },
    "debug": True
}

# 1. Salva in config.json
with open("config.json", "w", encoding="utf-8") as f:
    json.dump(config, f, indent=4, ensure_ascii=False)

print("✅ Configurazione salvata in config.json\n")

# 2. Leggi il file
with open("config.json", "r", encoding="utf-8") as f:
    config_letta = json.load(f)

print("📖 Configurazione caricata:")
print(f"  App: {config_letta['app_name']} v{config_letta['version']}")
print(f"  Debug: {config_letta['debug']}")
print(f"  Database: {config_letta['database']['name']} @ {config_letta['database']['host']}\n")

# 3. Modifica il valore di "debug"
config_letta["debug"] = False

# 4. Salva di nuovo
with open("config.json", "w", encoding="utf-8") as f:
    json.dump(config_letta, f, indent=4, ensure_ascii=False)

print("✏️  Debug mode disattivato\n")

# 5. Stampa la configurazione aggiornata
with open("config.json", "r", encoding="utf-8") as f:
    config_finale = json.load(f)

print("📋 Configurazione finale:")
print(json.dumps(config_finale, indent=2, ensure_ascii=False))

# Cleanup
Path("config.json").unlink()
````

**Spiegazione:**
- `json.dump(obj, file)` scrive un oggetto Python su file JSON
- `json.load(file)` legge un file JSON e lo converte in oggetto Python
- `indent=4` rende il JSON leggibile (formattato)
- `ensure_ascii=False` preserva caratteri accentati (àèéìòù)
- JSON è perfetto per configurazioni perché è leggibile e modificabile a mano

# 🎯 Sezione 3: Requests & API

## Esercizio 3.1: Scaricare Dati da API

**Obiettivo:** Scaricare e processare dati da un'API pubblica.

````python
# Soluzione Esercizio 3.1

import requests

# API: JSONPlaceholder (API fake per testing)
url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)

if response.ok:
    posts = response.json()  # Lista di dizionari
    
    # 1. Conta quanti post ci sono
    print(f"📊 Numero totale di post: {len(posts)}\n")
    
    # 2. Trova il post con il titolo più lungo
    post_piu_lungo = max(posts, key=lambda p: len(p["title"]))
    print(f"📝 Post con titolo più lungo:")
    print(f"   ID: {post_piu_lungo['id']}")
    print(f"   Titolo: {post_piu_lungo['title']}")
    print(f"   Lunghezza: {len(post_piu_lungo['title'])} caratteri\n")
    
    # 3. Stampa i primi 3 post
    print("🔝 Primi 3 post:")
    for post in posts[:3]:
        print(f"   [{post['id']}] {post['title']}")
else:
    print(f"❌ Errore: {response.status_code}")
````

**Spiegazione:**
- `requests.get(url)` fa una richiesta HTTP GET
- `response.ok` è `True` se lo status code è 2xx (successo)
- `response.json()` converte automaticamente la risposta JSON in lista/dizionario Python
- `max(..., key=lambda)` trova l'elemento con valore massimo secondo una funzione

## Esercizio 3.2: Gestire Errori HTTP

**Obiettivo:** Gestire correttamente gli errori nelle chiamate API.

````python
# Soluzione Esercizio 3.2

import requests

def scarica_utente(user_id):
    """
    Scarica i dati di un utente da JSONPlaceholder.
    
    Gestisce:
    - Status code 200: restituisce i dati
    - Status code 404: restituisce "Utente non trovato"
    - Altri errori: restituisce "Errore di connessione"
    """
    url = f"https://jsonplaceholder.typicode.com/users/{user_id}"
    
    try:
        response = requests.get(url, timeout=5)
        
        if response.status_code == 200:
            return response.json()
        elif response.status_code == 404:
            return "Utente non trovato"
        else:
            return f"Errore HTTP {response.status_code}"
    
    except requests.exceptions.RequestException as e:
        return "Errore di connessione"

# Test
print("Test 1 - Utente esistente (ID=1):")
utente1 = scarica_utente(1)
if isinstance(utente1, dict):
    print(f"  ✅ {utente1['name']} ({utente1['email']})\n")
else:
    print(f"  {utente1}\n")

print("Test 2 - Utente inesistente (ID=999):")
utente999 = scarica_utente(999)
print(f"  ⚠️  {utente999}\n")

print("Test 3 - URL invalido (ID=0):")
utente0 = scarica_utente(0)
print(f"  ⚠️  {utente0}")
````

**Spiegazione:**
- `response.status_code` contiene il codice HTTP (200=OK, 404=Not Found, 500=Server Error)
- `timeout=5` previene blocchi infiniti se il server non risponde
- `requests.exceptions.RequestException` cattura errori di rete (timeout, DNS, ecc.)
- Gestire gli errori rende l'applicazione robusta e user-friendly

## Esercizio 3.3: Parametri di Query

**Obiettivo:** Usare parametri per filtrare i risultati di un'API.

````python
# Soluzione Esercizio 3.3

import requests

# API: JSONPlaceholder comments
url = "https://jsonplaceholder.typicode.com/comments"

# 1. Scarica tutti i commenti del post con id=1
params = {"postId": 1}
response = requests.get(url, params=params)

if response.ok:
    commenti = response.json()
    
    # 2. Conta quanti commenti ha ricevuto
    print(f"📨 Il post #1 ha ricevuto {len(commenti)} commenti\n")
    
    # 3. Stampa l'email di chi ha scritto il primo commento
    if commenti:
        primo_commento = commenti[0]
        print(f"✉️  Primo commento da: {primo_commento['email']}")
        print(f"   Nome: {primo_commento['name']}")
        print(f"   Testo: {primo_commento['body'][:100]}...")
    
    # Bonus: mostra tutti i commenti
    print(f"\n📋 Tutti i commenti del post #1:")
    for i, comm in enumerate(commenti, 1):
        print(f"   {i}. {comm['name'][:40]}... ({comm['email']})")
else:
    print(f"❌ Errore: {response.status_code}")
````

**Spiegazione:**
- `params={"postId": 1}` viene convertito in `?postId=1` nell'URL
- Requests gestisce automaticamente l'encoding dei parametri
- Puoi passare più parametri: `params={"postId": 1, "limit": 10}`
- I parametri di query sono il modo standard per filtrare/paginare risultati API

# 🎯 Sezione 4: Pandas

## Esercizio 4.1: Creare e Filtrare DataFrame

**Obiettivo:** Creare un DataFrame e applicare filtri.

````python
# Soluzione Esercizio 4.1

import pandas as pd

# Crea un DataFrame con questi dati
dati = {
    "Nome": ["Alice", "Bob", "Charlie", "Diana", "Eve"],
    "Età": [25, 30, 35, 28, 42],
    "Città": ["Roma", "Milano", "Roma", "Napoli", "Milano"],
    "Stipendio": [30000, 45000, 50000, 35000, 60000]
}

df = pd.DataFrame(dati)
print("📊 DataFrame completo:")
print(df)
print()

# 1. Mostra solo le persone con età > 30
print("👥 Persone con età > 30:")
filtro_eta = df[df["Età"] > 30]
print(filtro_eta)
print()

# 2. Mostra solo le persone di Milano
print("🏙️  Persone di Milano:")
filtro_milano = df[df["Città"] == "Milano"]
print(filtro_milano)
print()

# 3. Mostra nome e stipendio delle persone di Roma
print("💰 Nome e stipendio delle persone di Roma:")
roma_subset = df[df["Città"] == "Roma"][["Nome", "Stipendio"]]
print(roma_subset)
print()

# 4. Calcola lo stipendio medio per città
print("📈 Stipendio medio per città:")
media_per_citta = df.groupby("Città")["Stipendio"].mean()
print(media_per_citta)
print()

# Bonus: statistiche complete per città
print("📊 Statistiche complete per città:")
stats = df.groupby("Città").agg({
    "Stipendio": ["mean", "min", "max"],
    "Nome": "count"
})
print(stats)
````

**Spiegazione:**
- `df[condizione]` filtra le righe che soddisfano la condizione (boolean indexing)
- `df[["col1", "col2"]]` seleziona solo alcune colonne (doppia parentesi quadra!)
- `groupby("colonna")` raggruppa i dati per categoria
- `.mean()`, `.sum()`, `.count()` sono funzioni di aggregazione
- `.agg()` permette di applicare più funzioni contemporaneamente

## Esercizio 4.2: Analisi CSV

**Obiettivo:** Caricare e analizzare un CSV (creato prima).

````python
# Soluzione Esercizio 4.2

import pandas as pd
from pathlib import Path

# Parte 1: Crea il CSV
vendite = {
    "Prodotto": ["Laptop", "Mouse", "Tastiera", "Monitor", "Laptop"],
    "Quantità": [2, 5, 3, 1, 1],
    "Prezzo": [999, 29, 79, 299, 999]
}

df = pd.DataFrame(vendite)
df.to_csv("vendite.csv", index=False)
print("✅ File vendite.csv creato\n")

# Parte 2: Analizza
# 1. Ricarica il CSV
df = pd.read_csv("vendite.csv")
print("📊 Dati caricati:")
print(df)
print()

# 2. Aggiungi colonna "Totale"
df["Totale"] = df["Quantità"] * df["Prezzo"]
print("💵 Con colonna Totale:")
print(df)
print()

# 3. Trova il prodotto con il totale più alto
idx_max = df["Totale"].idxmax()
prodotto_top = df.loc[idx_max]
print(f"🏆 Prodotto con totale più alto:")
print(f"   {prodotto_top['Prodotto']}: €{prodotto_top['Totale']:.2f}")
print()

# 4. Calcola il fatturato totale
fatturato_totale = df["Totale"].sum()
print(f"💰 Fatturato totale: €{fatturato_totale:.2f}")
print()

# 5. Raggruppa per Prodotto e somma le quantità vendute
print("📦 Quantità vendute per prodotto:")
vendite_per_prodotto = df.groupby("Prodotto").agg({
    "Quantità": "sum",
    "Totale": "sum"
})
print(vendite_per_prodotto)

# Cleanup
Path("vendite.csv").unlink()
````

**Spiegazione:**
- `pd.read_csv()` carica un CSV in un DataFrame
- `df["nuova_col"] = espressione` crea una nuova colonna calcolata
- `.idxmax()` restituisce l'indice della riga con valore massimo
- `.loc[indice]` accede a una riga specifica
- `groupby()` + `agg()` permette di aggregare dati duplicati

## Esercizio 4.3: Gestire Dati Mancanti

**Obiettivo:** Lavorare con valori NaN.

````python
# Soluzione Esercizio 4.3

import pandas as pd
import numpy as np
from pathlib import Path

# DataFrame con valori mancanti
dati = {
    "Nome": ["Mario", "Luigi", "Peach", "Toad"],
    "Età": [35, np.nan, 32, 28],
    "Email": ["mario@email.it", "luigi@email.it", np.nan, "toad@email.it"],
    "Punteggio": [100, 85, np.nan, 90]
}

df = pd.DataFrame(dati)
print("📊 DataFrame originale:")
print(df)
print()

# 1. Identifica quanti valori NaN ci sono per colonna
print("❓ Valori NaN per colonna:")
nan_count = df.isnull().sum()
print(nan_count)
print()

# 2. Sostituisci i NaN in "Età" con la media delle età
media_eta = df["Età"].mean()
df["Età"] = df["Età"].fillna(media_eta)
print(f"✏️  NaN in 'Età' sostituiti con la media ({media_eta:.1f})")
print(df)
print()

# 3. Sostituisci i NaN in "Email" con "non_disponibile@email.it"
df["Email"] = df["Email"].fillna("non_disponibile@email.it")
print("✏️  NaN in 'Email' sostituiti con email placeholder")
print(df)
print()

# 4. Rimuovi le righe con NaN in "Punteggio"
df_clean = df.dropna(subset=["Punteggio"])
print("🗑️  Righe con NaN in 'Punteggio' rimosse:")
print(df_clean)
print()

# 5. Salva il DataFrame pulito
df_clean.to_csv("dati_puliti.csv", index=False)
print("✅ DataFrame pulito salvato in 'dati_puliti.csv'")

# Verifica
print("\n🔍 Verifica: nessun NaN rimasto?")
print(df_clean.isnull().sum())

# Cleanup
Path("dati_puliti.csv").unlink()
````

**Spiegazione:**
- `.isnull()` restituisce un DataFrame di booleani (True dove c'è NaN)
- `.isnull().sum()` conta i NaN per colonna
- `.fillna(valore)` sostituisce i NaN con un valore specifico
- `.mean()` ignora automaticamente i NaN nel calcolo
- `.dropna(subset=["col"])` rimuove solo le righe con NaN in quella colonna
- La pulizia dei dati è fondamentale: molti algoritmi non accettano NaN
