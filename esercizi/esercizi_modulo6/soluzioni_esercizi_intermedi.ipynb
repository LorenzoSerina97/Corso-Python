# 🎯 Sezione 1: NumPy

## Esercizio 1.1: Array e Operazioni Vettoriali

**Obiettivo:** Creare array NumPy e confrontare performance con liste Python.

```python
# Soluzione Esercizio 1.1

import numpy as np
import time

# 1. Crea un array NumPy con i numeri da 0 a 999
arr = np.arange(1000000)
print(f"Array NumPy creato: {arr[:10]}... (primi 10 elementi)")
print(f"Lunghezza: {len(arr)}\n")

# 2. Calcola il quadrato di ogni elemento con NumPy
start_numpy = time.time()
quadrati_numpy = arr ** 2
end_numpy = time.time()
tempo_numpy = end_numpy - start_numpy

print(f"NumPy - Primi 10 quadrati: {quadrati_numpy[:10]}")
print(f"Tempo NumPy: {tempo_numpy:.6f} secondi\n")

# 3. Confronta con lista Python
lista = list(range(1000000))

start_lista = time.time()
quadrati_lista = [x**2 for x in lista]
end_lista = time.time()
tempo_lista = end_lista - start_lista

print(f"Lista Python - Primi 10 quadrati: {quadrati_lista[:10]}")
print(f"Tempo Lista: {tempo_lista:.6f} secondi\n")

# Confronto
print(f"📊 NumPy è {tempo_lista/tempo_numpy:.2f}x più veloce della lista Python!")
```

**Spiegazione:**
- `np.arange(1000)` crea un array con numeri da 0 a 999
- Le operazioni vettoriali NumPy (`arr ** 2`) sono molto più veloci delle list comprehension
- NumPy usa operazioni in C sotto il cofano, mentre Python usa loop interpretati
- Il modulo `time` permette di misurare le performance

## Esercizio 1.2: Array Multidimensionali

**Obiettivo:** Lavorare con matrici e operazioni su righe/colonne.

```python
# Soluzione Esercizio 1.2

import numpy as np

# 1. Crea una matrice 4x3 con numeri da 1 a 12
matrice = np.arange(1, 13).reshape(4, 3)
print("Matrice 4x3:")
print(matrice)
print(f"Forma: {matrice.shape}\n")

# 2. Accedi all'elemento in posizione (2, 1)
elemento = matrice[2, 1]
print(f"Elemento in posizione (2, 1): {elemento}\n")

# 3. Estrai la seconda riga completa
seconda_riga = matrice[1, :]
print(f"Seconda riga (indice 1): {seconda_riga}\n")

# 4. Calcola la somma di ogni colonna
somme_colonne = matrice.sum(axis=0)
print(f"Somma di ogni colonna: {somme_colonne}")
print(f"  Colonna 0: {somme_colonne[0]}")
print(f"  Colonna 1: {somme_colonne[1]}")
print(f"  Colonna 2: {somme_colonne[2]}\n")

# 5. Trasforma la matrice in un array 1D di 12 elementi
array_1d = matrice.reshape(12)
# Alternativa: array_1d = matrice.flatten()
print(f"Array 1D: {array_1d}")
print(f"Forma: {array_1d.shape}")
```

**Spiegazione:**
- `reshape(4, 3)` trasforma un array 1D in una matrice 4 righe x 3 colonne
- L'indicizzazione `[riga, colonna]` permette di accedere a elementi specifici
- `:` indica "tutte", quindi `[1, :]` significa "riga 1, tutte le colonne"
- `axis=0` applica l'operazione lungo le righe (producendo risultati per colonna)
- `axis=1` applicherebbe l'operazione lungo le colonne (risultati per riga)
- `reshape(12)` o `flatten()` trasformano una matrice in array 1D

```python
# Soluzione Esercizio 2.3

import pandas as pd

df = pd.DataFrame({
    "Nome": ["Mario", "Luigi", "Peach"],
    "Punti": [100, 85, 92],
    "Livello": [5, 4, 5]
})

print("DataFrame originale:")
print(df)
print()

# 1. Seleziona solo la colonna "Nome"
nomi = df["Nome"]
print("Colonna 'Nome':")
print(nomi)
print(f"Tipo: {type(nomi)}\n")  # pandas.Series

# 2. Seleziona le colonne "Nome" e "Punti"
subset = df[["Nome", "Punti"]]
print("Colonne 'Nome' e 'Punti':")
print(subset)
print(f"Tipo: {type(subset)}\n")  # pandas.DataFrame

# 3. Usa .loc per selezionare la prima riga
prima_riga = df.loc[0]
print("Prima riga (con .loc[0]):")
print(prima_riga)
print()

# 4. Usa .iloc per selezionare le prime 2 righe
prime_due_righe = df.iloc[0:2]
print("Prime 2 righe (con .iloc[0:2]):")
print(prime_due_righe)
print()

# 5. Seleziona il valore "Punti" di "Luigi" usando .loc
punti_luigi = df.loc[1, "Punti"]
print(f"Punti di Luigi: {punti_luigi}")
```

- `.loc[etichetta]` seleziona per etichetta/indice
- `.iloc[posizione]` seleziona per posizione numerica
- `.loc[riga, colonna]` seleziona un valore specifico
- `.iloc` usa slicing esclusivo come le liste Python: `[0:2]` = righe 0 e 1

# 🎯 Sezione 3: Manipolazione Dati

## Esercizio 3.1: Filtraggio Condizionale

**Obiettivo:** Applicare filtri multipli e complessi ai DataFrame.

```python
# Soluzione Esercizio 3.1

import pandas as pd

df = pd.DataFrame({
    "Prodotto": ["Laptop", "Mouse", "Tastiera", "Monitor", "Webcam"],
    "Prezzo": [999, 25, 75, 299, 89],
    "Quantità": [5, 50, 20, 10, 15],
    "Categoria": ["Elettronica", "Accessori", "Accessori", "Elettronica", "Accessori"]
})

print("DataFrame originale:")
print(df)
print()

# 1. Filtra prodotti con prezzo > 50 E quantità < 20
filtro1 = df[(df["Prezzo"] > 50) & (df["Quantità"] < 20)]
print("Prodotti con prezzo > 50 E quantità < 20:")
print(filtro1)
print()

# 2. Filtra prodotti Elettronica O con prezzo < 100
filtro2 = df[(df["Categoria"] == "Elettronica") | (df["Prezzo"] < 100)]
print("Prodotti Elettronica O con prezzo < 100:")
print(filtro2)
print()

# 3. Usa .isin() per trovare solo Laptop e Monitor
filtro3 = df[df["Prodotto"].isin(["Laptop", "Monitor"])]
print("Solo Laptop e Monitor:")
print(filtro3)
print()

# 4. Filtra prodotti il cui nome contiene "ta"
filtro4 = df[df["Prodotto"].str.contains("ta", case=False)]
print("Prodotti con 'ta' nel nome:")
print(filtro4)
```

**Spiegazione:**
- Le parentesi sono **obbligatorie** con `&` e `|` per evitare errori di precedenza
- `&` = AND logico (entrambe le condizioni devono essere vere)
- `|` = OR logico (almeno una condizione deve essere vera)
- `.isin([lista])` verifica se il valore appartiene alla lista
- `.str.contains()` cerca una sottostringa (case=False per ignorare maiuscole/minuscole)

## Esercizio 3.2: Creare Colonne Calcolate

**Obiettivo:** Aggiungere nuove colonne derivate da calcoli.

```python
# Soluzione Esercizio 3.2

import pandas as pd

df = pd.DataFrame({
    "Prodotto": ["Laptop", "Mouse", "Tastiera"],
    "Prezzo": [999, 25, 75],
    "Quantità": [5, 50, 20],
    "Sconto_%": [10, 0, 5]
})

print("DataFrame originale:")
print(df)
print()

# 1. Aggiungi colonna "Valore_Stock"
df["Valore_Stock"] = df["Prezzo"] * df["Quantità"]

# 2. Aggiungi colonna "Prezzo_Scontato"
df["Prezzo_Scontato"] = df["Prezzo"] * (1 - df["Sconto_%"] / 100)

# 3. Usa .apply() per creare "Categoria_Prezzo"
def classifica_prezzo(prezzo):
    if prezzo < 50:
        return "Economico"
    elif prezzo < 500:
        return "Medio"
    else:
        return "Premium"

df["Categoria_Prezzo"] = df["Prezzo"].apply(classifica_prezzo)

# 4. Crea colonna "Nome_Upper"
df["Nome_Upper"] = df["Prodotto"].str.upper()

print("DataFrame con nuove colonne:")
print(df)
```

**Spiegazione:**
- Le operazioni aritmetiche tra colonne sono vettoriali (elemento per elemento)
- `(1 - Sconto_%/100)` calcola il fattore di sconto (es. 10% → 0.90)
- `.apply(funzione)` applica una funzione personalizzata a ogni elemento
- Per logica complessa (if/elif/else), usare una funzione con apply
- `.str.upper()` converte stringhe in maiuscolo

## Esercizio 3.3: Boolean Masking Avanzato

**Obiettivo:** Combinare condizioni complesse per filtrare dati.

```python
# Soluzione Esercizio 3.3

import pandas as pd

df = pd.DataFrame({
    "Nome": ["Alice", "Bob", "Charlie", "Diana", "Eve"],
    "Età": [25, 30, 35, 28, 42],
    "Città": ["Roma", "Milano", "Roma", "Napoli", "Milano"],
    "Stipendio": [30000, 45000, 50000, 35000, 60000]
})

print("DataFrame originale:")
print(df)
print()

# 1. Trova persone di Roma con età > 26
filtro1 = df[(df["Città"] == "Roma") & (df["Età"] > 26)]
print("Persone di Roma con età > 26:")
print(filtro1)
print()

# 2. Trova persone con stipendio tra 35000 e 50000 (inclusi)
filtro2 = df[(df["Stipendio"] >= 35000) & (df["Stipendio"] <= 50000)]
print("Stipendio tra 35000 e 50000:")
print(filtro2)
print()

# 3. Trova persone di Milano O con stipendio > 50000
filtro3 = df[(df["Città"] == "Milano") | (df["Stipendio"] > 50000)]
print("Milano O stipendio > 50000:")
print(filtro3)
print()

# 4. Usa .query() per "Età > 30 and Città == 'Milano'"
filtro4 = df.query("Età > 30 and Città == 'Milano'")
print("Query: Età > 30 and Città == 'Milano':")
print(filtro4)
```

**Spiegazione:**
- Per intervalli, usare due condizioni: `(val >= min) & (val <= max)`
- `&` richiede entrambe le condizioni vere, `|` ne richiede almeno una
- `.query()` permette di scrivere condizioni come stringhe (più leggibile)
- In `.query()` si usa `and`/`or` invece di `&`/`|`
- Per stringhe in query, usare apici singoli all'interno: `"Città == 'Milano'"`

# 🎯 Sezione 4: Pulizia Dati

## Esercizio 4.1: Gestire Valori Mancanti

**Obiettivo:** Identificare e gestire NaN in modi diversi.

```python
# Soluzione Esercizio 4.1

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Nome": ["Alice", "Bob", np.nan, "Diana"],
    "Età": [25, np.nan, 35, 28],
    "Email": ["alice@email.it", "bob@email.it", "charlie@email.it", np.nan],
    "Stipendio": [30000, 45000, np.nan, 35000]
})

print("DataFrame originale:")
print(df)
print()

# 1. Conta quanti NaN ci sono per ogni colonna
print("NaN per colonna:")
print(df.isnull().sum())
print()

# 2. Rimuovi righe con NaN in "Nome"
df = df.dropna(subset=["Nome"])
print("Dopo rimozione righe con NaN in 'Nome':")
print(df)
print()

# 3. Sostituisci NaN in "Età" con la media
media_eta = df["Età"].mean()
df["Età"] = df["Età"].fillna(media_eta)
print(f"Media età: {media_eta:.2f}")
print("Dopo riempimento NaN in 'Età':")
print(df)
print()

# 4. Sostituisci NaN in "Email" con valore predefinito
df["Email"] = df["Email"].fillna("non_disponibile@email.it")
print("Dopo riempimento NaN in 'Email':")
print(df)
print()

# 5. Rimuovi colonna "Stipendio" se ha più di 1 NaN
nan_stipendio = df["Stipendio"].isnull().sum()
print(f"NaN in 'Stipendio': {nan_stipendio}")

if nan_stipendio > 1:
    df = df.drop(columns=["Stipendio"])
    print("Colonna 'Stipendio' rimossa (troppi NaN)")

print("\nDataFrame finale:")
print(df)
```

**Spiegazione:**
- `.isnull()` o `.isna()` creano una maschera booleana True/False per i valori mancanti
- `.dropna(subset=["Col"])` rimuove righe con NaN solo in quella colonna
- `.fillna(valore)` sostituisce tutti i NaN con un valore specifico
- `.mean()` calcola la media ignorando automaticamente i NaN
- Imputare con la media è una strategia comune per dati numerici

## Esercizio 4.2: Pulizia Stringhe

**Obiettivo:** Normalizzare e pulire dati testuali.

```python
# Soluzione Esercizio 4.2

import pandas as pd

df = pd.DataFrame({
    "Nome": ["  alice  ", "BOB", "Charlie ", " diana"],
    "Email": ["ALICE@EMAIL.IT", "bob@email.it  ", "Charlie@Email.It", "diana@email.it"],
    "Prezzo": ["€ 25,50", "€ 30,00", "€ 15,75", "€ 42,25"]
})

print("DataFrame originale:")
print(df)
print()

# 1. Rimuovi spazi extra da "Nome"
df["Nome"] = df["Nome"].str.strip()
print("Dopo rimozione spazi:")
print(df["Nome"])
print()

# 2. Converti "Nome" in formato Title Case
df["Nome"] = df["Nome"].str.title()
print("Dopo Title Case:")
print(df["Nome"])
print()

# 3. Converti "Email" in minuscolo e rimuovi spazi
df["Email"] = df["Email"].str.lower().str.strip()
print("Email pulite:")
print(df["Email"])
print()

# 4. Pulisci "Prezzo": rimuovi "€" e spazi, converti "," in "."
df["Prezzo"] = df["Prezzo"].str.replace("€", "").str.strip()
df["Prezzo"] = df["Prezzo"].str.replace(",", ".")
print("Prezzo pulito (ancora stringa):")
print(df["Prezzo"])
print()

# 5. Converti "Prezzo" in float
df["Prezzo"] = df["Prezzo"].astype(float)
print("DataFrame finale:")
print(df)
print(f"\nTipo di 'Prezzo': {df['Prezzo'].dtype}")
```

**Spiegazione:**
- `.str.strip()` rimuove spazi all'inizio e alla fine
- `.str.title()` mette la prima lettera in maiuscolo ("alice" → "Alice")
- `.str.lower()` converte tutto in minuscolo
- `.str.replace("old", "new")` sostituisce sottostringhe
- Si possono concatenare metodi: `.str.lower().str.strip()`
- `.astype(float)` converte una colonna in tipo numerico

## Esercizio 4.3: Conversione Tipi e Duplicati

**Obiettivo:** Convertire tipi di dato e rimuovere duplicati.

```python
# Soluzione Esercizio 4.3

import pandas as pd

df = pd.DataFrame({
    "ID": [1, 2, 2, 3, 4],
    "Età": ["25", "30", "30", "35", "abc"],
    "Data": ["2024-01-15", "2024-02-20", "2024-02-20", "2024-03-10", "2024-04-05"]
})

print("DataFrame originale:")
print(df)
print()

# 1. Identifica righe duplicate
duplicati = df.duplicated()
print("Righe duplicate (True = duplicato):")
print(duplicati)
print(f"Numero duplicati: {duplicati.sum()}\n")

# 2. Rimuovi righe duplicate complete
df = df.drop_duplicates()
print("Dopo rimozione duplicati:")
print(df)
print()

# 3. Converti "Età" in int, gestendo errori con errors='coerce'
df["Età"] = pd.to_numeric(df["Età"], errors="coerce")
print("Dopo conversione 'Età' in numerico:")
print(df)
print(f"Tipo di 'Età': {df['Età'].dtype}\n")

# 4. Converti "Data" in datetime
df["Data"] = pd.to_datetime(df["Data"])
print("Dopo conversione 'Data' in datetime:")
print(df)
print(f"Tipo di 'Data': {df['Data'].dtype}\n")

# 5. Conta quanti NaN sono stati creati dalla conversione di "Età"
nan_eta = df["Età"].isnull().sum()
print(f"NaN creati in 'Età': {nan_eta}")
print("(Il valore 'abc' è stato convertito in NaN)")
```

**Spiegazione:**
- `.duplicated()` restituisce True per righe duplicate (eccetto la prima occorrenza)
- `.drop_duplicates()` rimuove tutte le righe duplicate
- `pd.to_numeric(errors='coerce')` converte in numero, mettendo NaN dove fallisce
- `pd.to_datetime()` converte stringhe in oggetti datetime
- I valori non convertibili ("abc") diventano NaN con `errors='coerce'`

# 🎯 Sezione 5: Analisi & GroupBy

## Esercizio 5.1: GroupBy Base

**Obiettivo:** Raggruppare dati e calcolare statistiche aggregate.

```python
# Soluzione Esercizio 5.1

import pandas as pd

df = pd.DataFrame({
    "Categoria": ["Elettronica", "Accessori", "Elettronica", "Accessori", "Elettronica"],
    "Prodotto": ["Laptop", "Mouse", "Monitor", "Tastiera", "Webcam"],
    "Vendite": [1200, 50, 800, 75, 150],
    "Quantità": [5, 20, 8, 15, 10]
})

print("DataFrame originale:")
print(df)
print()

# 1. Raggruppa per "Categoria" e calcola la somma delle "Vendite"
vendite_per_categoria = df.groupby("Categoria")["Vendite"].sum()
print("Vendite totali per categoria:")
print(vendite_per_categoria)
print()

# 2. Raggruppa per "Categoria" e calcola la media della "Quantità"
media_quantita = df.groupby("Categoria")["Quantità"].mean()
print("Quantità media per categoria:")
print(media_quantita)
print()

# 3. Conta quanti prodotti ci sono per categoria
conteggio = df.groupby("Categoria").size()
print("Numero prodotti per categoria:")
print(conteggio)
print()

# 4. Trova le vendite massime per categoria
vendite_max = df.groupby("Categoria")["Vendite"].max()
print("Vendite massime per categoria:")
print(vendite_max)
```

**Spiegazione:**
- `.groupby("Col")` raggruppa i dati per valori univoci in quella colonna
- Dopo groupby, si seleziona la colonna da aggregare: `["Vendite"]`
- `.sum()`, `.mean()`, `.max()` applicano la funzione a ogni gruppo
- `.size()` conta il numero di righe in ogni gruppo
- Il risultato è una Series con l'indice = valori della colonna di raggruppamento

## Esercizio 5.2: GroupBy Avanzato

**Obiettivo:** Usare aggregazioni multiple con agg().

```python
# Soluzione Esercizio 5.2

import pandas as pd

df = pd.DataFrame({
    "Città": ["Roma", "Milano", "Roma", "Napoli", "Milano"],
    "Nome": ["Alice", "Bob", "Charlie", "Diana", "Eve"],
    "Stipendio": [30000, 45000, 35000, 28000, 50000],
    "Età": [25, 30, 35, 28, 42]
})

print("DataFrame originale:")
print(df)
print()

# 1. Raggruppa per "Città" e calcola sum, mean, count per "Stipendio"
stats_stipendio = df.groupby("Città")["Stipendio"].agg(["sum", "mean", "count"])
print("Statistiche stipendio per città:")
print(stats_stipendio)
print()

# 2. Usa .agg() con dizionario per funzioni diverse
stats_multiple = df.groupby("Città").agg({
    "Stipendio": ["mean", "max"],
    "Età": ["min", "max"]
})
print("Statistiche multiple per città:")
print(stats_multiple)
print()

# 3. Trova la città con lo stipendio medio più alto
stipendi_medi = df.groupby("Città")["Stipendio"].mean()
citta_top = stipendi_medi.idxmax()
stipendio_top = stipendi_medi.max()

print(f"Città con stipendio medio più alto: {citta_top}")
print(f"Stipendio medio: €{stipendio_top:.2f}")
```

**Spiegazione:**
- `.agg(["func1", "func2"])` applica più funzioni alla stessa colonna
- Con dizionario `.agg({"Col1": ["f1"], "Col2": ["f2"]})` si applicano funzioni diverse a colonne diverse
- `.idxmax()` restituisce l'indice (etichetta) del valore massimo
- Il risultato di `.agg()` su più colonne è un DataFrame multi-indice

## Esercizio 5.3: Merge DataFrame

**Obiettivo:** Unire due DataFrame con merge().

```python
# Soluzione Esercizio 5.3

import pandas as pd

vendite = pd.DataFrame({
    "Prodotto_ID": [1, 2, 3, 4],
    "Quantità": [10, 5, 8, 12],
    "Prezzo": [999, 25, 299, 75]
})

prodotti = pd.DataFrame({
    "ID": [1, 2, 3, 5],
    "Nome": ["Laptop", "Mouse", "Monitor", "Tastiera"],
    "Categoria": ["Elettronica", "Accessori", "Elettronica", "Accessori"]
})

print("DataFrame vendite:")
print(vendite)
print()

print("DataFrame prodotti:")
print(prodotti)
print()

# 1. Inner join tra vendite e prodotti
inner_join = pd.merge(vendite, prodotti, left_on="Prodotto_ID", right_on="ID")
print("Inner join:")
print(inner_join)
print()

# 2. Left join (mantieni tutte le vendite)
left_join = pd.merge(vendite, prodotti, left_on="Prodotto_ID", right_on="ID", how="left")
print("Left join:")
print(left_join)
print()

# 3. Quante righe ha l'inner join? E il left join? Perché?
print(f"Righe inner join: {len(inner_join)}")
print(f"Righe left join: {len(left_join)}")
print()
print("Spiegazione:")
print("- Inner join ha 3 righe (prodotti 1, 2, 3 hanno match in entrambe le tabelle)")
print("- Left join ha 4 righe (mantiene tutte le vendite, prodotto 4 non ha match)")
print("- Prodotto 5 in 'prodotti' non appare in nessuno perché non ha vendite")
print()

# 4. Aggiungi colonna "Totale" al risultato
left_join["Totale"] = left_join["Quantità"] * left_join["Prezzo"]
print("Left join con colonna Totale:")
print(left_join)
```

**Spiegazione:**
- `pd.merge(df1, df2, left_on="Col1", right_on="Col2")` unisce due DataFrame
- `how="inner"` (default): solo righe con match in entrambi
- `how="left"`: tutte le righe da sinistra + match da destra (NaN se manca)
- `how="right"`: tutte le righe da destra + match da sinistra
- `how="outer"`: tutte le righe da entrambi
- Il merge è simile alle JOIN in SQL
